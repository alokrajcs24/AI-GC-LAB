```python

import random
import math

N = 8
INITIAL_TEMP = 5.0
COOLING_RATE = 0.99
MAX_STEPS = 1500
MAX_RESTARTS = 100

random.seed(42)


# Count every attacking pair of queens
def attacking_pairs(board):
    pairs = []

    for i in range(N):
        for j in range(i + 1, N):

            if board[i] == board[j]:
                pairs.append(f"Q{i+1}-Q{j+1} (same row)")

            elif abs(board[i] - board[j]) == abs(i - j):
                pairs.append(f"Q{i+1}-Q{j+1} (diagonal)")

    return pairs


# Cost = total number of attacking pairs
def cost(board):
    return len(attacking_pairs(board))


# Display the board as an 8 x 8 grid
def print_board(board):
    print("     ", end="")
    for col in range(N):
        print(col + 1, end=" ")
    print()

    print("    " + "-" * 17)

    for row in range(N):
        print(f"{row + 1:2} | ", end="")

        for col in range(N):
            if board[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")

        print()

    print("    " + "-" * 17)


def show_board_details(title, board):
    pairs = attacking_pairs(board)

    print(f"\n{title}")
    print("Cost =", len(pairs))
    print_board(board)

    if pairs:
        print("Attacking pairs:", ", ".join(pairs))
    else:
        print("Attacking pairs: None")


def simulated_annealing():

    for restart in range(1, MAX_RESTARTS + 1):

        # Random permutation: one queen per row and column
        board = list(range(N))
        random.shuffle(board)

        current_cost = cost(board)
        temperature = INITIAL_TEMP

        print("\n" + "=" * 55)
        print("RESTART:", restart)

        show_board_details("INITIAL BOARD", board)

        if current_cost == 0:
            print("\nSOLUTION FOUND!")
            show_board_details("FINAL SOLUTION", board)
            return

        for step in range(1, MAX_STEPS + 1):

            print("\n" + "=" * 55)
            print(f"RESTART {restart} | STEP {step}")
            print(f"Temperature T = {temperature:.6f}")

            # Show current state
            show_board_details("CURRENT BOARD", board)

            # Generate one neighboring state by swapping two queens
            col1, col2 = random.sample(range(N), 2)

            candidate = board.copy()
            candidate[col1], candidate[col2] = (
                candidate[col2], candidate[col1]
            )

            candidate_cost = cost(candidate)

            # Show proposed state
            print(
                f"\nMove: Swap the row positions of "
                f"Q{col1 + 1} and Q{col2 + 1}"
            )

            show_board_details("PROPOSED NEIGHBOR", candidate)

            # Calculate energy difference
            delta = candidate_cost - current_cost

            # Calculate probability of acceptance
            if delta <= 0:
                probability = 1.0
            else:
                probability = math.exp(-delta / temperature)

            random_number = random.random()

            print("\nCOST CALCULATION")
            print("Current cost =", current_cost)
            print("Candidate cost =", candidate_cost)
            print("Delta E = Candidate cost - Current cost")
            print("Delta E =", delta)
            print("Temperature =", round(temperature, 6))
            print("Acceptance probability =", round(probability, 6))
            print("Random number =", round(random_number, 6))

            # Accept better/equal states or probabilistically accept worse ones
            if delta <= 0 or random_number < probability:

                board = candidate
                current_cost = candidate_cost

                print("DECISION: ACCEPTED")
                print("New current cost =", current_cost)

            else:
                print("DECISION: REJECTED")
                print("Board remains unchanged.")
                print("Current cost =", current_cost)

            # Goal test
            if current_cost == 0:
                print("\n" + "#" * 55)
                print("SOLUTION FOUND!")
                print("Restart =", restart)
                print("Step =", step)

                show_board_details("FINAL SOLUTION", board)

                print("Final cost = 0")
                print("No queens attack each other.")
                return

            # Reduce temperature
            temperature *= COOLING_RATE

        print("\nMaximum steps reached. Starting a new restart.")

    print("\nCould not find a solution within the restart limit.")
    print("Try increasing MAX_RESTARTS.")


simulated_annealing()
```
