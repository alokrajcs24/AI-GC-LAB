```python

from heapq import heappush, heappop
from itertools import count

GOAL = (1, 2, 3, 4, 5, 6, 7, 8, 0)

# Count misplaced tiles, excluding blank (0)
def h(state):
    return sum(1 for i in range(9)
               if state[i] != 0 and state[i] != GOAL[i])

# Generate all valid moves of the blank
def neighbors(state):
    result = []
    zero = state.index(0)
    r, c = divmod(zero, 3)

    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, name in moves:
        nr, nc = r + dr, c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            pos = nr * 3 + nc
            new_state = list(state)
            new_state[zero], new_state[pos] = (
                new_state[pos], new_state[zero]
            )
            result.append((name, tuple(new_state)))

    return result

def show(state):
    for i in range(0, 9, 3):
        print(" ".join("_" if x == 0 else str(x)
                       for x in state[i:i+3]))

def astar(start):
    pq = []
    serial = count()

    g_score = {start: 0}
    parent = {start: None}
    move_taken = {start: None}
    closed = set()

    heappush(pq, (h(start), next(serial), start))

    print("INITIAL STATE")
    show(start)
    print("g =", 0, "h =", h(start), "f =", h(start))

    step = 0

    while pq:
        f_value, _, current = heappop(pq)

        if current in closed:
            continue

        closed.add(current)
        g = g_score[current]
        step += 1

        print("\nSTEP", step, "- SELECTED NODE")
        show(current)
        print("g(n) =", g)
        print("h(n) =", h(current))
        print("f(n) =", g + h(current))

        if current == GOAL:
            print("\nGOAL REACHED!")
            break

        print("\nCHILD NODES:")

        for move, nxt in neighbors(current):
            new_g = g + 1

            if nxt in closed:
                continue

            if new_g < g_score.get(nxt, float("inf")):
                g_score[nxt] = new_g
                parent[nxt] = current
                move_taken[nxt] = move

                hn = h(nxt)
                fn = new_g + hn

                heappush(pq, (fn, next(serial), nxt))

                print("\nMove blank:", move)
                show(nxt)
                print("g =", new_g,
                      "h =", hn,
                      "f =", fn)

    else:
        print("No solution found.")
        return

    # Reconstruct the solution
    path = []
    node = GOAL

    while node is not None:
        path.append(node)
        node = parent[node]

    path.reverse()

    print("\n========== SOLUTION PATH ==========")

    for i, state in enumerate(path):
        print("\nMove", i)
        show(state)
        print("g =", i,
              "h =", h(state),
              "f =", i + h(state))

        if i > 0:
            print("Blank moved:", move_taken[state])

    print("\nTotal moves =", len(path) - 1)


# Initial state
start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

astar(start)
```
