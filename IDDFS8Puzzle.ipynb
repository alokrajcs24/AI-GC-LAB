```python

GOAL = (1, 2, 3, 4, 5, 6, 7, 8, 0)

def show(state):
    for i in range(0, 9, 3):
        print(" ".join("_" if x == 0 else str(x)
                       for x in state[i:i+3]))

def get_neighbors(state):
    result = []
    zero = state.index(0)
    r, c = divmod(zero, 3)

    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, name in moves:
        nr, nc = r + dr, c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            pos = nr * 3 + nc
            new_state = list(state)
            new_state[zero], new_state[pos] = (
                new_state[pos], new_state[zero]
            )
            result.append((name, tuple(new_state)))

    return result

def dls(state, depth, limit, path, moves):
    if state == GOAL:
        return path, moves

    if depth == limit:
        return None

    for move, nxt in get_neighbors(state):
        if nxt not in path:
            result = dls(
                nxt, depth + 1, limit,
                path + [nxt], moves + [move]
            )

            if result is not None:
                return result

    return None

def iddfs(start, max_depth=50):
    for limit in range(max_depth + 1):
        print("\nDepth limit =", limit)

        result = dls(start, 0, limit, [start], [])

        if result is not None:
            path, moves = result

            print("\nSOLUTION FOUND!")
            for i, state in enumerate(path):
                print("\nStep", i)
                show(state)

                if i > 0:
                    print("Blank moved:", moves[i-1])

            print("\nTotal moves =", len(path) - 1)
            return

    print("No solution found within maximum depth.")

start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

print("Initial State:")
show(start)

iddfs(start)
```
