```python

# Vacuum Cleaner Agent: Simple Reflex, Goal-Based and Model-Based

def simple_reflex(location, status):
    if status == "Dirty":
        return "Suck"
    elif location == "A":
        return "Right"
    else:
        return "Left"


def goal_based(location, rooms):
    if rooms["A"] == "Dirty":
        return "Suck" if location == "A" else "Left"

    if rooms["B"] == "Dirty":
        return "Suck" if location == "B" else "Right"

    return "Goal Achieved"


def model_based(location, status, model):
    model[location] = status

    if model["A"] == "Dirty":
        return "Suck" if location == "A" else "Left"

    if model["B"] == "Dirty":
        return "Suck" if location == "B" else "Right"

    return "Goal Achieved"


def run_agent(agent_name, agent_function):
    rooms = {"A": "Dirty", "B": "Dirty"}
    model = {"A": "Unknown", "B": "Unknown"}
    location = "A"

    print("\n---", agent_name, "---")

    for step in range(5):
        status = rooms[location]

        if agent_name == "Simple Reflex":
            action = agent_function(location, status)
        elif agent_name == "Goal-Based":
            action = agent_function(location, rooms)
        else:
            action = agent_function(location, status, model)

        print("Location:", location,
              "| Rooms:", rooms,
              "| Action:", action)

        if action == "Suck":
            rooms[location] = "Clean"
        elif action == "Right":
            location = "B"
        elif action == "Left":
            location = "A"
        else:
            print("Goal achieved! Both rooms are clean.")
            break


run_agent("Simple Reflex", simple_reflex)
run_agent("Goal-Based", goal_based)
run_agent("Model-Based", model_based)
```
