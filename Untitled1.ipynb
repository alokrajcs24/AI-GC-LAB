```python
import random

board = [" " for _ in range(9)]

def display_board():
    print()
    print(board[0] + " | " + board[1] + " | " + board[2])
    print("--+---+--")
    print(board[3] + " | " + board[4] + " | " + board[5])
    print("--+---+--")
    print(board[6] + " | " + board[7] + " | " + board[8])
    print()

def check_winner(player):
    winning_positions = [
        [0, 1, 2],
        [3, 4, 5],
        [6, 7, 8],
        [0, 3, 6],
        [1, 4, 7],
        [2, 5, 8],
        [0, 4, 8],
        [2, 4, 6]
    ]

    for position in winning_positions:
        if all(board[i] == player for i in position):
            return True

    return False

def computer_move():
    empty = [i for i in range(9) if board[i] == " "]

    for i in empty:
        board[i] = "O"
        if check_winner("O"):
            return

        board[i] = " "

    for i in empty:
        board[i] = "X"
        if check_winner("X"):
            board[i] = "O"
            return

        board[i] = " "

    if 4 in empty:
        board[4] = "O"
        return

    corners = [0, 2, 6, 8]
    available_corners = [i for i in corners if i in empty]

    if available_corners:
        board[random.choice(available_corners)] = "O"
        return

    board[random.choice(empty)] = "O"

print("===== TIC-TAC-TOE =====")
print("You = X")
print("Computer = O")

while True:

    display_board()

    try:
        move = int(input("Enter position (1-9): ")) - 1

        if move < 0 or move > 8:
            print("Enter a number between 1 and 9.")
            continue

        if board[move] != " ":
            print("Position already occupied.")
            continue

        board[move] = "X"

    except ValueError:
        print("Please enter a number.")
        continue

    if check_winner("X"):
        display_board()
        print("You WIN!")
        break

    if " " not in board:
        display_board()
        print("DRAW!")
        break

    computer_move()

    if check_winner("O"):
        display_board()
        print("Computer WINS!")
        break

    if " " not in board:
        display_board()
        print("DRAW!")
        break
```
