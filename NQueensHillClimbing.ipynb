```python

def cost(board):
    conflicts = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j]:
                conflicts += 1
            elif abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def print_grid(board):
    n = len(board)

    print("  ", end="")
    for col in range(n):
        print(col + 1, end=" ")
    print()

    for row in range(n):
        print(row + 1, end=" ")

        for col in range(n):
            if board[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")

        print()


def hill_climbing():
    n = 4
    board = [0, 0, 0, 0]
    step = 1

    while True:
        current_cost = cost(board)

        print("\n========== STEP", step, "==========")
        print("Current Board | Cost =", current_cost)
        print_grid(board)

        if current_cost == 0:
            print("\nSOLUTION FOUND!")
            print("Final Cost = 0")
            return

        best_cost = current_cost
        best_board = None
        candidate_no = 1

        # Move one queen at a time
        for col in range(n):
            for row in range(n):
                if row == board[col]:
                    continue

                candidate = board.copy()
                candidate[col] = row
                candidate_cost = cost(candidate)

                print("\nCandidate", candidate_no)
                print("Move: Queen in column", col + 1,
                      "to row", row + 1)
                print("Cost =", candidate_cost)
                print_grid(candidate)

                candidate_no += 1

                if candidate_cost < best_cost:
                    best_cost = candidate_cost
                    best_board = candidate

        if best_board is None:
            print("\nLocal minimum reached. Restarting...")
            board = [0, 1, 2, 3]

            import random
            random.shuffle(board)
        else:
            board = best_board
            print("\nSelected Next Board | Cost =", best_cost)
            print_grid(board)

        step += 1


hill_climbing()
```
